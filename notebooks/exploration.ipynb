```python

features = pd.DataFrame([extract_features(x) for x in our_df["URL"]])
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

mask = y.notna()
X = features.loc[mask].reset_index(drop=True)
y_clean = y.loc[mask].astype(int).reset_index(drop=True)

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_clean,
    test_size=0.2,
    random_state=42,
    stratify=y_clean
)

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

pca = PCA(n_components=0.95)

X_train_pca = pca.fit_transform(X_train_scaled)
X_test_pca = pca.transform(X_test_scaled)
```

```python
import joblib

joblib.dump(scaler, "../models/scaler.pkl")
joblib.dump(pca, "../models/pca.pkl")
```

```python
pca_columns = [f"PC{i+1}" for i in range(X_train_pca.shape[1])]

X_train_pca_df = pd.DataFrame(
    X_train_pca,
    columns=pca_columns
)

X_test_pca_df = pd.DataFrame(
    X_test_pca,
    columns=pca_columns
)

X_train_pca_df.to_csv("x_train_pca.csv", index=False)
X_test_pca_df.to_csv("x_test_pca.csv", index=False)

y_train.to_csv("y_train.csv", index=False)
y_test.to_csv("y_test.csv", index=False)
```

```python
print("Original features:", X_train.shape[1])
print("PCA features:", X_train_pca.shape[1])
print("Variance retained:", pca.explained_variance_ratio_.sum())
```
